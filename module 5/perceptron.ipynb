```python
import numpy as np
```

```python
w = np.array([2, -3])
b = -4


def neyron(x):
    z = np.dot(x, w) + b
    return int(z > 0)


print(f"[5, 1] -> {neyron([5, 1])}")
print(f"[3, 2] -> {neyron([2, 3])}")
print(f"[3, 3] -> {neyron([3, 3])}")
print(f"[6, 1] -> {neyron([6, 1])}")
```

```python
w = np.array([2, -3])
b = -4


def neyron(x):
    z = np.dot(x, w) + b
    return [int(z > 0), z]


y_pred = neyron([4, 2])[0]
z = neyron([4, 2])[1]

print(f"[4, 2] -> y={y_pred}, z={z}")
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])


def gate(w, b):
    return (X @ np.array(w) + b > 0).astype(int)


print(f"AND -> {gate([2, 2], -3)}")
print(f"OR -> {gate([2, 2], -1)}")
print(f"OR -> {gate([2, 2], 1)}")
print(f"NAND -> {gate([22, -2], 3)}")
```

```python
class PerceptronNP:
    def __init__(self, lr=0.5, epochs=50):
        self.lr, self.epochs = lr, epochs

    def predict(self, x):
        return (x @ self.w + self.b > 0).astype(int)

    def fit(self, X, y):
        self.w, self.b = np.zeros(X.shape[1]), 0.0
        for ep in range(1, self.epochs + 1):
            xato = 0
            for xi, yi in zip(X, y):
                upd = self.lr * (yi - self.predict(xi))
                self.w, self.b = self.w + upd * xi, self.b + upd
                xato += int(upd != 0.0)
            if xato == 0:
                return ep
        return self.epochs
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

p = PerceptronNP()

print(f"epochs count -> {p.fit(X, [0, 0, 0, 1])}")
print(f"w -> {p.w}, \nb -> {p.b}")
```

```python
import seaborn as sns
from sklearn.model_selection import train_test_split

belgilar = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
df = sns.load_dataset("penguins").dropna(subset=belgilar)
print(df.shape)

d = df[df["species"].isin(["Adelie", "Gentoo"])]
X = d[belgilar].values
y = (d["species"] == "Gentoo").astype(int).values

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
print(len(Xtr), len(Xte))
```

```python
from sklearn.linear_model import Perceptron

model = Perceptron(max_iter=1000, random_state=42)
model.fit(Xtr, ytr)

print(model.score(Xte, yte))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix

pipe = make_pipeline(StandardScaler(), Perceptron(random_state=42))

pipe.fit(Xtr, ytr)
print(pipe.score(Xte, yte))
print(confusion_matrix(yte, pipe.predict(Xte)))
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score

cv = StratifiedKFold(5, shuffle=True, random_state=42)

p0 = Perceptron(random_state=42)
raw = cross_val_score(p0, X, y, cv=cv)
sc = cross_val_score(pipe, X, y, cv=cv)

print(raw.mean().round(4))
print(sc.mean().round(4))
```
