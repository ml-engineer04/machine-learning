```python
z = np.array([-4, -2, 0, 2, 4], dtype=float)
sigmoid = 1 / (1 + np.exp(-z))

print(sigmoid.round(4))
print((sigmoid * (1 - sigmoid)).round(4))
print(np.tanh(z).round(4))
```

```python
import torch
import torch.nn.functional as F

for f in (F.relu, F.leaky_relu):
    w = torch.tensor(1.0, requires_grad=True)
    h = f(w * 2 - 5)
    ((h - 1) ** 2).backward()
    print(round(h.item(), 4), round(w.grad.item(), 4))
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

for a, b in ((2, 0), (6, -12)):
    p = a * x + b
    print(a, b, np.mean((y - p) ** 2), np.mean(np.abs(y - p)))
```

```python
def softmax(logits: np.ndarray, axis: int = -1) -> np.ndarray:
    x = np.asarray(logits, dtype=np.float64)
    shifted = x - np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(shifted)
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)


scores = np.array([[2.0, 1.0, 0.1], [1000.0, 1001.0, 999.0]])
probs = softmax(scores, axis=-1)
print(probs)
```
