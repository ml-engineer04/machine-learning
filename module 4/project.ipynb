```python
df["target"] = (df["income"] == ">50K").astype(int)
```

```python
df = df.drop(columns=["income"])
```

```python
df = df.drop_duplicates()
```

```python
print(f"Shape: {df.shape}")
```

```python
columns = df.columns.tolist()
print(columns)
```

```python
df[columns] = df[columns].replace("?", "NA")
```

```python
y = df["target"]
X = df.drop(columns=["target"])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.2, stratify=y
)

# X_train, X_valid, y_train, y_valid = train_test_split(
#     X_train, y_train, test_size=0.2, random_state=42
# )
```

```python
nums = X_train.select_dtypes(include="number").columns.tolist()
cats = X_train.select_dtypes(exclude="number").columns.tolist()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from catboost import CatBoostClassifier

nums_pipe = Pipeline(
    [("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())]
)

cats_pipe = Pipeline(
    [
        ("imp", SimpleImputer(strategy="most_frequent")),
        (
            "oh",
            OneHotEncoder(handle_unknown="ignore"),
        ),
    ]
)

pre = ColumnTransformer([("num", nums_pipe, nums), ("cat", cats_pipe, cats)])

pipe = Pipeline([("preprocessor", pre), ("model", CatBoostClassifier(verbose=0))])
```

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
import time
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_validate
from sklearn.pipeline import Pipeline

logistic_mode = Pipeline(
    [
        ("preprocessor", pre),
        ("model", LogisticRegression(max_iter=1000, random_state=42)),
    ]
)

scoring_metrics = {"f1": "f1", "roc_auc": "roc_auc"}

start = time.perf_counter()
logistic_cv = cross_validate(
    logistic_mode, X_train, y_train, cv=cv, scoring=scoring_metrics, n_jobs=-1
)
logistic_time = time.perf_counter() - start

print(f"F1 per fold: {logistic_cv['test_f1']}")
print(
    f"Mean F1: {logistic_cv['test_f1'].mean():.2f} (+/- {logistic_cv['test_f1'].std():.2f})"
)

print(f"ROC AUC per fold: {logistic_cv['test_roc_auc']}")
print(
    f"Mean ROC AUC: {logistic_cv['test_roc_auc'].mean():.2f} (+/- {logistic_cv['test_roc_auc'].std():.2f})"
)

print(f"logistic_time -> {logistic_time:.4f}s")
```

```python
from xgboost import XGBClassifier

xgb_model = Pipeline(
    [
        ("preprocessor", pre),
        (
            "model",
            XGBClassifier(
                n_estimators=200,
                learning_rate=0.1,
                max_depth=6,
                random_state=42,
                eval_metric="logloss",
                n_jobs=-1,
            ),
        ),
    ]
)

scoring_metrics = {"f1": "f1", "roc_auc": "roc_auc"}

start = time.perf_counter()
xgb_cv = cross_validate(
    xgb_model, X_train, y_train, cv=cv, scoring=scoring_metrics, n_jobs=-1
)
xgb_time = time.perf_counter() - start

print(f"F1 per fold: {xgb_cv['test_f1']}")
print(f"Mean F1: {xgb_cv['test_f1'].mean():.2f} (+/- {xgb_cv['test_f1'].std():.2f})")

print(f"ROC AUC per fold: {xgb_cv['test_roc_auc']}")
print(
    f"Mean ROC AUC: {xgb_cv['test_roc_auc'].mean():.2f} (+/- {xgb_cv['test_roc_auc'].std():.2f})"
)

print(f"xgb_time -> {xgb_time:.4f}s")
```

```python
from lightgbm import LGBMClassifier

lgbm_model = Pipeline(
    [
        ("preprocessor", pre),
        (
            "model",
            LGBMClassifier(
                n_estimators=200,
                learning_rate=0.1,
                max_depth=-1,
                num_leaves=31,
                random_state=42,
                n_jobs=-1,
                verbosity=-1,
            ),
        ),
    ]
)

scoring_metrics = {"f1": "f1", "roc_auc": "roc_auc"}

start = time.perf_counter()
lgbm_cv = cross_validate(
    lgbm_model, X_train, y_train, cv=cv, scoring=scoring_metrics, n_jobs=-1
)
lgbm_time = time.perf_counter() - start

print(f"F1 per fold: {lgbm_cv['test_f1']}")
print(f"Mean F1: {lgbm_cv['test_f1'].mean():.2f} (+/- {lgbm_cv['test_f1'].std():.2f})")

print(f"ROC AUC per fold: {lgbm_cv['test_roc_auc']}")
print(
    f"Mean ROC AUC: {lgbm_cv['test_roc_auc'].mean():.2f} (+/- {lgbm_cv['test_roc_auc'].std():.2f})"
)

print(f"lgbm_time -> {lgbm_time:.4f}s")
```

```python
from catboost import CatBoostClassifier

catboost_model = Pipeline(
    [
        ("preprocessor", pre),
        (
            "model",
            CatBoostClassifier(
                iterations=200,
                learning_rate=0.1,
                depth=6,
                random_seed=42,
                verbose=0,
                thread_count=1,
            ),
        ),
    ]
)

scoring_metrics = {"f1": "f1", "roc_auc": "roc_auc"}

start = time.perf_counter()
catboost_cv = cross_validate(
    catboost_model, X_train, y_train, cv=cv, scoring=scoring_metrics, n_jobs=-1
)
catboost_time = time.perf_counter() - start

print(f"F1 per fold: {catboost_cv['test_f1']}")
print(
    f"Mean F1: {catboost_cv['test_f1'].mean():.2f} (+/- {catboost_cv['test_f1'].std():.2f})"
)

print(
    f"Mean ROC AUC: {catboost_cv['test_roc_auc'].mean():.2f} (+/- {catboost_cv['test_roc_auc'].std():.2f})"
)

print(f"catboost_time -> {catboost_time:.4f}s")
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline(
    [
        ("preprocessor", pre),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                max_depth=None,
                min_samples_split=2,
                min_samples_leaf=1,
                random_state=42,
                n_jobs=1,
            ),
        ),
    ]
)

scoring_metrics = {"f1": "f1", "roc_auc": "roc_auc"}

start = time.perf_counter()
rf_cv = cross_validate(
    rf_model, X_train, y_train, cv=cv, scoring=scoring_metrics, n_jobs=-1
)
rf_time = time.perf_counter() - start

print(f"F1 per fold: {rf_cv['test_f1']}")
print(f"Mean F1: {rf_cv['test_f1'].mean():.2f} (+/- {rf_cv['test_f1'].std():.2f})")

print(f"ROC AUC per fold: {rf_cv['test_roc_auc']}")
print(
    f"Mean ROC AUC: {rf_cv['test_roc_auc'].mean():.2f} (+/- {rf_cv['test_roc_auc'].std():.2f})"
)

print(f"rf_time -> {rf_time:.4f}s")
```

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
)


def metrics(X_test, y_test, model):
    y_pred = model.predict(X_test)
    y_probs = model.predict_proba(X_test)[:, 1]

    result = {
        "Accuracy": round(accuracy_score(y_test, y_pred), 2),
        "Precision": round(precision_score(y_test, y_pred), 2),
        "Recall": round(recall_score(y_test, y_pred), 2),
        "F1 Score": round(f1_score(y_test, y_pred), 2),
        "ROC AUC": round(roc_auc_score(y_test, y_probs), 2),
    }

    return pd.DataFrame(result, index=[0])
```
